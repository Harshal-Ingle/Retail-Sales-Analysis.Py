```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("retail_sales.csv")
df.head()
```

```python
#Q1.1 Load the dataset and convert it into a NumPy array.

arr = df.to_numpy()
arr
```

```python
#Q1.2. Find the total revenue generated across all stores.

totalRevenue = df["Total_Sales"].sum()
print("Total revenue across all stores:", totalRevenue)
```

```python
#Q1.3 Identify the highest and lowest price per unit among all products.

maxPPU = df["Price_Per_Unit"].max()
minPPU = df["Price_Per_Unit"].min()
print("Highest price per unit among all products:", maxPPU)
print("Lowest price per unit among all products:", minPPU)
```

```python
#Q1.4 Calculate the average number of units sold per product.

avgUnitsSold = df["Units_Sold"].mean()
print("Average number of units sold per product:", avgUnitsSold)
```

```python
#Q2.1 Determine the store with the highest total revenue.

totalRevenuePerStore = df.groupby("Store_ID")["Total_Sales"].sum()
maxRevenueStoreID = totalRevenuePerStore.idxmax()
print("Store ID with highest total revenus:",maxRevenueStoreID)
```

```python
#Q2.2 Identify the store with the lowest number of units sold.

totalUnitsSoldPerStore = df.groupby("Store_ID")["Units_Sold"].sum()
minUnitsStoreID = totalUnitsSoldPerStore.idxmin()
print("Store ID with lowest number of units sold:",minUnitsStoreID)
```

```python
#Q2.3 Compute the standard deviation of total sales across all stores.

standardDevAcrossStores = df["Total_Sales"].std()
print("Standard Deviation across all stores:", standardDevAcrossStores)
```
